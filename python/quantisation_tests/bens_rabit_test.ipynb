```python
vals = [0, 1, 2, 3]
MAX_VAL = 3
```

```python
def make_grid(dimensions):
    grids= []
    current = [0 for _ in range(dimensions)]
    
    return np.array(extend_grid(grids, current)) - 1.5
    
def extend_grid(grids, current):
    
    finished = False
    
    
    while not finished:
        grids.append(current)
        
        if np.any(current != MAX_VAL):
            current = get_next_grid(current)                
        else:
            finished = True
            
    return grids
        
def get_next_grid(vec):
   index = 0
   finished = False
   
   new_vec = np.copy(vec)
   
   while not finished:
        new_vec[index] += 1
        
        if new_vec[index] == MAX_VAL + 1 and index < len(new_vec):
            new_vec[index] = 0
        else:
            return new_vec
            
        index += 1
    
```

```python
rabbit_codebook = make_grid(3)
normalised_rabbit_codebook = rabbit_codebook / np.linalg.norm(rabbit_codebook, axis=1, keepdims=True)
```

```python
old_vector = np.array([-0.045782919051816966, -0.4880434399430473, -0.8716177631574817])
new_vector = np.array([-0.7571879625320435, -0.5521767139434814, 0.34895169734954834])
extra_vector = l2_normalise(np.asarray([0.2,0.1,0.3]))
```

```python
old_rabit_id = np.argmax(np.dot(normalised_rabbit_codebook, old_vector))
new_rabit_id = np.argmax(np.dot(normalised_rabbit_codebook, new_vector))
extra_rabit_id = np.argmax(np.dot(normalised_rabbit_codebook, extra_vector))

print(old_rabit_id)
print(new_rabit_id)
print(extra_rabit_id)
```
