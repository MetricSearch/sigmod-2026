```python
def get_next_normed_random_vec( d: int ) -> np.array:
    """
    :param d: the number of dimensions required
    :return: a l2 normed random vector of dimension d and
    """
    return l2_normalise(np.random.normal(0, 1, (d,)))
```

```python
def print_elems(vector1: np.array, vector2: np.array) -> str:
    for i in range(len(vector1)):
        if vector1[i] != vector2[i]:
            print( f"Different in element {i} : {vector1[i]} {vector2[i]}")
```

```python
from move import move_on_unit_sphere, get_random_orthogonal_vector

def move_and_test(vector: np.array, angle: float):
    old_evp = get_evp_row_arg_partition_stable(vector)
    old_rabit = get_rabit(vector)
    old_vector = vector
    evp_diff = False
    rabit_diff = False

    np.set_printoptions(precision=10, suppress=False, linewidth=np.inf)

    for i in range(10000):
        direction = get_random_orthogonal_vector(vector)
        vector = move_on_unit_sphere(vector, direction, angle)             # Generate the moved vector

        evp = get_evp_row_arg_partition_stable(vector)
        rabit = get_rabit(vector)

        if not np.array_equal(old_evp, evp):
            print(f"Evps differ iter {i} {old_evp} {evp}")
            print(f"Vector: {old_vector} {vector}")
            print_elems(old_vector,vector)
            evp_diff = True

        if not np.array_equal(old_rabit, rabit):
            print(f"Rabit differ iter {i} {old_rabit} {rabit}")
            print(f"Vector: {old_vector} {vector}")
            print_elems(old_vector,vector)
            rabit_diff = True

        if evp_diff or rabit_diff:
            break

        old_evp = evp
        old_rabit = rabit
        old_vector = vector
00
```
